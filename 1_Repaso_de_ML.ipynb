## <font color="#de9f68">**1.2. La limitación de la Linealidad y la Monotonía**

Los modelos de <font color="#adc261">**regresión y clasificación lineal**</font> (como la regresión Softmax) asumen transformaciones afines de la forma:

$$\mathbf{o} = \mathbf{X} \mathbf{W} + \mathbf{b}$$

Esta formulación lineal impone la condición de monotonicidad en cada variable de forma independiente. Esto significa que <font color="#adc261">**un incremento en una coordenada de entrada $x_i$ debe traducirse en un cambio constante en la salida del modelo**</font> (**positivo o negativo**, dependiendo puramente de la dirección del peso $w_i$), <font color="#adc261">**asumiendo todas las demás variables constantes.**</font>

En problemas reales, esta suposición es geométricamente inadecuada:

* <font color="#91b9e3">**Falta de monotonía física:**</font> Al predecir el <font color="#adc261">**estado de salud**</font> en función de la <font color="#adc261">**temperatura corporal**</font>, tanto una temperatura menor a 37 °C (hipotermia) como una mayor (fiebre) <font color="#adc261">**aumentan el riesgo del paciente**</font>.

    > Un modelo lineal estricto <font color="#a091e3">**no puede representar esta parábola sin una transformación previa por diseño**</font> (como medir la distancia absoluta $|T - 37|$).

* <font color="#91b9e3">**Dependencia del contexto espacial**</font>: En <font color="#adc261">**procesamiento de imágenes**</font> (e.g., distinguir gatos de perros), la importancia de un píxel individual no es monótona; <font color="#adc261">**su significado semántico cambia drásticamente en función de los valores de las coordenadas de los píxeles circundantes**</font>.

    > <font color="#a091e3">**Invertir la luminosidad de una imagen preserva su contenido semántico, lo cual rompe por completo las premisas lineales.**

## <font color="#de9f68">**1.3. El Teorema del Colapso Lineal**

> <font color="#a091e3">**Apilar capas de neuronas lineales no incrementa la capacidad de representación de una red profunda.**</font>

Consideremos una red con una capa oculta sin funciones de activación no lineales:

$$\mathbf{H} = \mathbf{X} \mathbf{W}^{(1)} + \mathbf{b}^{(1)}$$ $$\mathbf{O} = \mathbf{H} \mathbf{W}^{(2)} + \mathbf{b}^{(2)}$$

Al sustituir la primera ecuación en la segunda, **podemos colapsar el sistema mediante simple álgebra de matrices:**

$$\mathbf{O} = (\mathbf{X} \mathbf{W}^{(1)} + \mathbf{b}^{(1)})\mathbf{W}^{(2)} + \mathbf{b}^{(2)}$$ $$\mathbf{O} = \mathbf{X} (\mathbf{W}^{(1)}\mathbf{W}^{(2)}) + (\mathbf{b}^{(1)} \mathbf{W}^{(2)} + \mathbf{b}^{(2)})$$ $$\mathbf{O} = \mathbf{X} \mathbf{W}{\text{eq}} + \mathbf{b}{\text{eq}}$$

donde $\mathbf{W}{\text{eq}} = \mathbf{W}^{(1)}\mathbf{W}^{(2)}$ y $\mathbf{b}{\text{eq}} = \mathbf{b}^{(1)} \mathbf{W}^{(2)} + \mathbf{b}^{(2)}$.

> <font color="#a091e3">**Geométricamente, apilar capas lineales consecutivas no altera la topología del espacio de características. La frontera de decisión final del modelo siempre se reduce a un hiperplano plano y rígido en $\mathbb{R}^d$, limitando la expresividad del modelo.**</font>

## <font color="#de9f68">**1.4. Deformadores Espaciales: Inyectando No Linealidad**</font>

Para <font color="#adc261">**permitir**</font> que un <font color="#adc261">**modelo**</font> ajuste <font color="#adc261">**límites de decisión altamente no lineales, introducimos funciones de activación no lineales**</font> $\sigma(\cdot)$ **aplicadas de forma individual** (elemento a elemento) tras cada transformación afín:

$$\mathbf{H} = \sigma(\mathbf{X} \mathbf{W}^{(1)} + \mathbf{b}^{(1)})$$

> <font color="#a091e3">**Estas funciones tuercen, estiran y pliegan el espacio de características original de modo que geometrías complejas que no son linealmente separables en el espacio original se tornen linealmente separables en el espacio latente transformado.**</font>

* <font color="#91b9e3">**ReLU (Rectified Linear Unit):**</font> $\operatorname{ReLU}(x) = \max(0, x)$. Actúa como un operador lineal a trozos. Conserva la linealidad original para valores positivos y proyecta de forma ortogonal todos los valores negativos sobre el cero. Permite modelar funciones continuas complejas mediante la aproximación por segmentos planos plegados.
    
* <font color="#91b9e3">**Sigmoide**</font>: $\sigma(x) = \frac{1}{1 + e^{-x}}$. Comprime el dominio real de valores $(-\infty, \infty)$ al intervalo acotado $(0, 1)$, ideal para modelar distribuciones de probabilidad de Bernoulli. Geométricamente, aproxima suavemente una función escalón.

* <font color="#91b9e3">**Tanh (Tangente Hiperbólica)**</font>: $\operatorname{tanh}(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}} = \frac{1 - e^{-2x}}{1 + e^{-2x}}$. Comprime los datos al rango $(-1, 1)$, exhibiendo simetría respecto al origen. Alrededor de cero, se comporta de forma aproximadamente lineal.

### <font color="#e391e3">**1.5.1. Aprendizaje Supervisado: Fronteras de Decisión y Separabilidad**</font>

En el aprendizaje supervisado, <font color="#adc261">**cada coordenada de entrada $\mathbf{x}_i$ posee una etiqueta asociada $y_i$**</font>. El objetivo es encontrar una frontera geométrica continua que separe las clases.

* <font color="#91b9e3">**Modelos Paramétricos (e.g., Redes Neuronales / MLPs):**</font> El <font color="#adc261">**optimizador actualiza los pesos**</font> de forma iterativa para <font color="#adc261">**deformar gradualmente el espacio a través de capas ocultas sucesivas**</font>, hasta que las <font color="#adc261">**representaciones se vuelven linealmente separables en la capa de salida.**</font>

* <font color="#91b9e3">**Modelos No Paramétricos (e.g., $k$-Nearest Neighbors o $1$-NN):**</font> El <font color="#adc261">**clasificador no define una frontera analítica global**</font>. En su lugar, <font color="#adc261">**memoriza la ubicación exacta de las muestras de entrenamiento y subdivide el espacio localmente**</font> en "vecindades de decisión" mediante métricas de distancia:

$$d(\mathbf{x}, \mathbf{x}') = \sqrt{\sum_{k=1}^d (x_k - x'_k)^2}$$

> Este tipo de modelo <font color="#a091e3">**se expande de forma natural conforme aumenta el volumen de datos de entrenamiento.**

#### <font color="#d6a487">**1.5.2.1. PCA (Principal Component Analysis): Rotación y Proyección Ortogonal**

PCA calcula las direcciones en las que la distribución de datos exhibe la mayor varianza. Algebraicamente, <font color="#adc261">**se resuelve encontrando los autovectores de la matriz de covarianza de los datos centralizados:**

$$\mathbf{\Sigma} = \frac{1}{n}\mathbf{X}^T\mathbf{X}$$

Geométricamente, <font color="#adc261">**PCA rota de manera ortogonal los ejes coordenados originales hacia un nuevo sistema de coordenadas donde el primer eje**</font> (primer componente principal) <font color="#adc261">**apunta a lo largo de la elipse de máxima dispersión de los datos**</font>. Proyectar sobre estos componentes <font color="#adc261">**permite reducir la dimensionalidad eliminando los ejes con menor varianza estructural.**</font>


#### <font color="#d6a487">**1.5.2.2. $K$-Means y las Celdas de Voronoi**

El algoritmo $K$-Means <font color="#adc261">**posiciona de forma iterativa $K$ centroides $\{\mathbf{\mu}_1, \dots, \mathbf{\mu}_k\}$ en el espacio de características.**</font> En cada iteración, <font color="#adc261">**asigna cada punto al centroide más cercano utilizando la distancia euclídea.**

Este proceso particiona el espacio en regiones cerradas llamadas **Celdas de Voronoi** (o polígonos de Voronoi). Matemáticamente, la celda $V_i$ asociada al centroide $\mathbf{\mu}_i$ se define como:

$$V_i = \{ \mathbf{x} \in \mathbb{R}^d \mid d(\mathbf{x}, \mathbf{\mu}_i) \leq d(\mathbf{x}, \mathbf{\mu}_j) \quad \forall j \neq i \}$$

> Las <font color="#a091e3">**fronteras entre celdas adyacentes son hiperplanos perpendiculares equidistantes a los dos centroides correspondientes.**


### <font color="#e391e3">**1.5.3. Máquinas de Vectores de Soporte (SVM): La Geometría del Margen Máximo**

El objetivo de un SVM lineal es <font color="#adc261">**encontrar un hiperplano separador óptimo de la forma:**

$$\mathbf{w}^T \mathbf{x} + b = 0$$

que <font color="#adc261">**maximice la distancia mínima (el margen) hacia los puntos más cercanos de cada clase (conocidos como **Vectores de Soporte**).**

```
                  Frontera de Margen (+1)
                     o      o      /
                            x     /      o
                           /     /
  Clase A (o)             /     /              Clase B (x)
                         /     /
             Vector ───> o    /    x <─── Vector de Soporte
             de Soporte /    /
                       /    /     x
                      /    /
                     /   Hiperplano de Decisión (wᵀx + b = 0)
                    /
         Frontera de Margen (-1)
```

#### <font color="#d6a487">**1.5.3.1. Formulación del Margen Canónico**

Para <font color="#adc261">**asegurar una separación nítida con la mayor distancia de seguridad de ambas clases, imponemos las restricciones**:

$$\mathbf{w}^T \mathbf{x}_i + b \geq +1 \quad \text{si } y_i = +1$$
$$\mathbf{w}^T \mathbf{x}_i + b \leq -1 \quad \text{si } y_i = -1$$

Estas condiciones se unifican de manera elegante multiplicando por la etiqueta de clase $y_i \in \{-1, +1\}$:

$$y_i (\mathbf{w}^T \mathbf{x}_i + b) \geq 1 \quad \forall i \in \{1, \dots, n\}$$

#### <font color="#d6a487">**1.5.3.2. Maximización del Margen y Minimización de la Norma $\|\mathbf{w}\|$**

La <font color="#adc261">**distancia geométrica entre las dos fronteras de margen paralelas de separación**</font> ($\mathbf{w}^T\mathbf{x} + b = 1$ y $\mathbf{w}^T\mathbf{x} + b = -1$) está definida por la <font color="#adc261">**proyección en la dirección normal al hiperplano**:

$$\text{Margen} = \frac{2}{\|\mathbf{w}\|}$$

Para maximizar este margen de separación, debemos <font color="#adc261">**minimizar la norma del vector de pesos**</font> $\|\mathbf{w}\|$. Este problema se plantea convencionalmente como una <font color="#adc261">**tarea de optimización cuadrática convexa**:

$$\min_{\mathbf{w}, b} \frac{1}{2} \|\mathbf{w}\|^2 \quad \text{sujeto a: } y_i (\mathbf{w}^T \mathbf{x}_i + b) \geq 1 \quad \forall i$$

Esta formulación matemática es el equivalente directo a la **regularización $\ell_2$ (Weight Decay)** empleada en redes neuronales, <font color="#adc261">**la cual restringe la norma de los pesos para evitar que el modelo genere fronteras oscilantes propensas al sobreajuste.**

#### **1.5.3.3. La Formulación Dual de Lagrange y la Aparición del Producto Interno**

Para incorporar las restricciones directas, <font color="#adc261">**definimos el Lagrangiano Primal con multiplicadores de Lagrange $\alpha_i \geq 0$:**

$$\mathcal{L}(\mathbf{w}, b, \boldsymbol{\alpha}) = \frac{1}{2} \|\mathbf{w}\|^2 - \sum_{i=1}^n \alpha_i \left[ y_i (\mathbf{w}^T \mathbf{x}_i + b) - 1 \right]$$

Si calculamos las derivadas parciales de $\mathcal{L}$ con respecto a los parámetros primales $\mathbf{w}$ y $b$, e igualamos a cero, obtenemos las condiciones de optimalidad:

1.  $$\frac{\partial \mathcal{L}}{\partial \mathbf{w}} = 0 \implies \mathbf{w} = \sum_{i=1}^n \alpha_i y_i \mathbf{x}_i$$

2.  $$\frac{\partial \mathcal{L}}{\partial b} = 0 \implies \sum_{i=1}^n \alpha_i y_i = 0$$

Al sustituir estos resultados de vuelta en el Lagrangiano, transformamos el problema en su **Formulación Dual**:

$$\max_{\boldsymbol{\alpha}} \sum_{i=1}^n \alpha_i - \frac{1}{2} \sum_{i=1}^n \sum_{j=1}^n \alpha_i \alpha_j y_i y_j \langle \mathbf{x}_i, \mathbf{x}_j \rangle$$

sujeto a:

$$\alpha_i \geq 0 \quad \forall i, \quad \text{y} \quad \sum_{i=1}^n \alpha_i y_i = 0$$

> <font color="#a091e3">**Relevancia Geométrica del Producto Interno: Toda la optimización y predicción del clasificador SVM se rige exclusivamente por el término $\langle \mathbf{x}_i, \mathbf{x}_j \rangle$, que representa el producto interno espacial entre las muestras del conjunto de entrenamiento.**
